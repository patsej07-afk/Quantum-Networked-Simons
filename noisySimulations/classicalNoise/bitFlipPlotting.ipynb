```python
import matplotlib
import numpy as np
from matplotlib import pyplot as plt
```

```python
matplotlib.rc('text', usetex=True)
matplotlib.rc('font', family='serif')
BIG_TEXT = 12
MEDIUM_TEXT = 10
SMALL_TEXT = 8
plt.rcParams['figure.dpi'] = 150
plt.rcParams['axes.axisbelow'] = True
plt.rcParams.update({'font.size': SMALL_TEXT})
```

```python
P_ARRAY = np.arange(0,0.105,0.005)
N_ARRAY = np.arange(2,16,1)[::-1]
x_ticks = np.arange(0,len(P_ARRAY),1)
y_ticks = np.arange(0,len(N_ARRAY),1)
```

```python
fig = plt.figure(figsize=(3.45, 2.5), layout='constrained')
with open('data/bitflipIterations.txt', 'r') as file:
    numTrials = int(file.read())
successHeatmap = np.load('data/bitflipSuccessHeatmap.npy')
successHeatmap /= numTrials
plt.imshow(successHeatmap[::-1])
plt.xlabel(r'$p$', fontsize=MEDIUM_TEXT)
plt.ylabel(r'$n$', fontsize=MEDIUM_TEXT)
plt.xticks(x_ticks, P_ARRAY, rotation=45, ha='right')
plt.yticks(y_ticks, N_ARRAY)
plt.colorbar()
plt.savefig('../../figures/bitFlipModel.png')
```
